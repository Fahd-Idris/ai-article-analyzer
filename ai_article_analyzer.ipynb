```python
import torch # Needed for handling tensors (e.g., softmax, argmax)
from transformers import pipeline
from transformers import AutoTokenizer, AutoModelForSequenceClassification

print('='.center(25,'='))
print('AI ARTICLE ANALYZER'.center(25,' '))
print('='.center(25,'='))
text = input('Enter text here: ')

# Initialize sentiment analysis pipeline
sentiment_pipeline = pipeline("sentiment-analysis")
sentiment_result = sentiment_pipeline(text)

checkpoint="distilbert-base-uncased-finetuned-sst-2-english"
tokenizer = AutoTokenizer.from_pretrained(checkpoint)
model = AutoModelForSequenceClassification.from_pretrained(checkpoint)

# 1. Correct model input: Tokenize the text to get input_ids and attention_mask
inputs = tokenizer(text, return_tensors="pt")

# 2. Pass the prepared inputs to the model
model_outputs = model(**inputs)

# 3. Process model outputs to get the predicted label
logits = model_outputs.logits
probabilities = torch.softmax(logits, dim=1)
predicted_class_idx = torch.argmax(probabilities, dim=1).item()
predicted_label = model.config.id2label[predicted_class_idx]

# Get raw tokens for display (as string list)
raw_tokens = tokenizer.tokenize(text)

# Decode the input_ids back to text for 'human readable interpretation'
# This shows how the model 'sees' the text after tokenization
decoded_input_text_from_ids = tokenizer.decode(inputs['input_ids'][0], skip_special_tokens=True)


print('1. Sentiment (pipeline) : ' + f"Label: {sentiment_result[0]['label']}, Score: {sentiment_result[0]['score']:.4f}")
print('2. Raw Tokens : ' + str(raw_tokens)) # Convert list of tokens to string for printing
print('3. Model Output (Predicted Label) : ' + predicted_label) # Print the predicted label from AutoModelForSequenceClassification
print('4. Decoded Input Text : ' + decoded_input_text_from_ids) # Print the input text decoded from its token IDs
```
